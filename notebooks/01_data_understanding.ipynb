# Data Understanding

**Objective:**
Load the Online Retail dataset and perform an initial inspection to understand its structure, missing values, duplicates, and general properties before proceeding to deeper Exploratory Data Analysis (EDA).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration
pd.set_option('display.max_columns', None)
```

## 1. Data Loading

```python
file_path = '../data/raw/Online_Retail.xlsx'
df = pd.read_excel(file_path)
print("Dataset loaded successfully.")
```

## 2. Structure Inspection

```python
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
df.info()
```

## 3. First Rows

```python
df.head()
```

## 4. Missing Values & Duplicates

```python
missing_values = df.isnull().sum()
missing_percentages = (missing_values / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing_values, 'Percentage (%)': missing_percentages})
print("--- Missing Values ---")
print(missing_df[missing_df['Missing Count'] > 0])

duplicates = df.duplicated().sum()
print(f"\n--- Duplicates ---\nNumber of duplicate rows: {duplicates}")
```

## 5. Statistical Summary

```python
df.describe(include='all')
```

### Key Findings
1. **Structure**: The dataset contains 541,909 rows and 8 columns.
2. **Data Types**: The dataset contains categorical (`InvoiceNo`, `StockCode`, `Description`, `Country`), numerical (`Quantity`, `UnitPrice`, `CustomerID`), and temporal (`InvoiceDate`) features.
3. **Missing Values**:
    * `CustomerID`: 135,080 missing values (~24.93%). This is a critical issue as customer segmentation requires customer identification.
    * `Description`: 1,454 missing values (~0.27%).
4. **Duplicates**: There are 5,268 duplicate rows in the dataset.
5. **Unusual Values**: There are negative quantities and potentially zero or negative UnitPrices, which indicate returns or bad records.

### Decisions Made
* Proceed to EDA to investigate the exact distributions of Quantity, UnitPrice, and the temporal patterns.
* Missing `CustomerID` records will likely need to be excluded or handled separately in the cleaning phase since they cannot be mapped to a specific customer for segmentation.
* Duplicate rows will need to be dropped in the preprocessing pipeline.
